# Torch import diagnostic
Use this notebook to compare the current Jupyter kernel and the Isaac Sim Python environment.

- Cell 1 shows interpreter info and paths for the current kernel.
- Cell 2 tries to import torch and reports the exception if it fails.
- Cell 3 runs a short script with the Isaac Sim launcher (omni_python/python.sh) and prints its sys.path.

```python
import os
import sys
from pprint import pprint

print('Interpreter:', sys.executable)
print('Version:', sys.version.split()[0])
print('PYTHONPATH:', os.environ.get('PYTHONPATH'))
print('LD_LIBRARY_PATH:', os.environ.get('LD_LIBRARY_PATH'))
print('First sys.path entries:')
pprint(sys.path[:8])
```

```python
import importlib
import traceback

try:
    torch = importlib.import_module('torch')
    print('torch version:', torch.__version__)
except Exception as exc:
    print(f'torch import failed: {exc.__class__.__name__}: {exc}')
    traceback.print_exc()
```

```python
import json
import os
import shlex
import subprocess
from pathlib import Path


candidates = []
env_launcher = os.environ.get('OMNI_PYTHON', '').strip()
if env_launcher:
    candidates.append(Path(env_launcher))

candidates.extend([
    Path('/isaac-sim/kit/python/bin/omni_python'),
    Path('/isaac-sim/omni_python'),
    Path('/isaac-sim/python.sh'),
])

omni = next((path for path in candidates if path.exists()), None)
if not omni:
    raise FileNotFoundError('No Isaac Sim python launcher found; update candidates list above.')

cmd = [str(omni), '-c', 'import sys, json, torch;\nprint(sys.executable);\nprint(torch.__version__);\nprint(json.dumps(sys.path, indent=2))']
print('Running:', ' '.join(shlex.quote(c) for c in cmd))
output = subprocess.check_output(cmd, text=True)
print('--- launcher output start ---')
print(output)
print('--- launcher output end ---')
```
